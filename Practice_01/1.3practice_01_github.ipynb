## 4. Push the changes to GitHub 🖥️
Git works in three stages: **working directory → staging area (`add`) → local repository (`commit`) → remote repository (`push`)**.

**Option 1. Terminal** (inside the repository folder):
```bash
git status
git add README.md
git commit -m "Update README with project description"
git push
```

**Option 2. VS Code:** *Source Control* → write the message → **Commit** → **Push** (↑ arrow in the *Graph* bar, or **⋯ → Push**).

> ℹ️ Run `push` from the terminal or VS Code, not from this notebook: the notebook cannot answer the login prompt.

````python
# ▶️ Check the state of your repository
!git status
````

**How to read it:**
- `Your branch is up to date with 'origin/main'` → everything is on GitHub ✅
- `Your branch is ahead of 'origin/main' by N commits` → you committed but **did not push** yet.
- `Changes not staged` / `Untracked files` → there are changes without `add` + `commit`.

### 5.1 Keep the repository clean with `.gitignore`
Some files should never be uploaded: Jupyter checkpoints, virtual environments, secrets, or large datasets. The next cell adds the most common ones to your `.gitignore` (it only adds what is missing).

````python
# ▶️ Add common entries to .gitignore (safe to run several times)
import subprocess
from pathlib import Path

root = Path(subprocess.run(["git", "rev-parse", "--show-toplevel"],
                           capture_output=True, text=True).stdout.strip() or ".")
gitignore = root / ".gitignore"
wanted = [".ipynb_checkpoints/", "__pycache__/", ".venv/", "venv/", ".env", ".DS_Store", "*.pkl", "data/raw/"]

current = gitignore.read_text().splitlines() if gitignore.exists() else []
missing = [w for w in wanted if w not in current]
if missing:
    with gitignore.open("a") as f:
        f.write("\n# Added in practice 1\n" + "\n".join(missing) + "\n")
print(f"{gitignore}: added {missing if missing else 'nothing (already complete)'}")
````

## Troubleshooting
| Message | Cause | Solution |
|---|---|---|
| `Password authentication is not supported` | You typed your GitHub password | Use VS Code login, `gh auth login`, or a token (step 2.1) |
| `Repository not found` | Wrong repository name, or the token cannot see a private repo | Copy the URL from **Code → HTTPS**; check the token's repository access |
| `403 … Permission denied` | Token without write permission | Classic token with **`repo`**, or fine-grained with **Contents: Read and write** |
| `nothing to commit, working tree clean` | The file was not saved or has no changes | Save (Cmd/Ctrl + S) and run `git status` again |
| `error: pathspec '#' did not match` | You pasted `# comments` in zsh | Type the commands without comments |
| `fatal: not a git repository` | You are outside the cloned folder | `cd` into the repository folder |
| `rejected … fetch first` | GitHub has changes you don't have locally | `git pull`, then `git push` |

## ✅ Deliverable
Fill in your repository URL and run the final check.

````python
# ✍️ Your repository URL
REPO_URL = "https://github.com/<your-user>/<your-repo>"
````

````python
# ▶️ Final check
import subprocess
from pathlib import Path

def git(*args):
    r = subprocess.run(["git", *args], capture_output=True, text=True)
    return r.stdout.strip() if r.returncode == 0 else ""

subprocess.run(["git", "fetch", "--quiet"], capture_output=True)
root = Path(git("rev-parse", "--show-toplevel") or ".")
remote = git("remote", "get-url", "origin")
ahead = git("rev-list", "--count", "@{u}..HEAD")
readme = root / "README.md"

checks = {
    "1. Repository URL filled in": "<your-user>" not in REPO_URL,
    "2. Local copy cloned from GitHub": "github.com" in remote,
    "2. URL matches the remote": REPO_URL.rstrip("/").removesuffix(".git").lower() in remote.lower(),
    "3. README has more than the default content": readme.exists() and len(readme.read_text(encoding="utf-8").splitlines()) > 3,
    "4. All commits pushed to GitHub": ahead == "0",
    "5. Practice_01/ folder exists": (root / "Practice_01").is_dir(),
}
for name, ok in checks.items():
    print(("✅" if ok else "❌"), name)
print("\nCommits:", git("rev-list", "--count", "HEAD"), "| Remote:", remote or "not found")
````
